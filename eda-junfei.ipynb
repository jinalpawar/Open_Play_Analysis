```python
def parse_time(s):
    return int(s[0:4]), int(s[5:7]), int(s[8:10]), int(s[11:13]), int(s[14:16])
parse_time("2025-07-24T08:00:00Z")
```

```python
import datetime
 
```

```python
def get_play_distribution_in_one_platform(player_distribution, data):# an empty dict, and a dataframe of one platform's data

    adjust = {}
    for _ in range(len(pid_list)):
        adjust[pid_list[_]] = time_adjust[_]
 
    if 'approximate_session_start' in data.columns:
        start_col, end_col = 'approximate_session_start', 'approximate_session_end'
    else:
        start_col, end_col = 'session_start', 'session_end'

    minutes_col = 'minutes' if 'minutes' in data.columns else None

    pids = data['pid'].tolist()
    starts = data[start_col].tolist()
    ends = data[end_col].tolist()
    mins = data[minutes_col].tolist() if minutes_col is not None else None

    for i in range(len(pids)):
        pid = pids[i]
        y, m, d, h, mi = parse_time(str(starts[i]))
        wd = datetime.date(y, m, d).weekday()

        if mins is not None:
            remaining = float(mins[i])
        else:
            y2, m2, d2, h2, mi2 = parse_time(str(ends[i]))
            remaining = (h2 - h) * 60 + (mi2 - mi)
            if d2 != d:                                 
                remaining += 24 * 60
        h = h + adjust[pid]
        if h < 0:
            h += 24
            wd = (wd - 1) % 7
        elif h >= 24:
            h -= 24
            wd = (wd + 1) % 7

        while remaining > 0:
            chunk = min(60 - mi, remaining)
            week = 1 if wd >= 5 else 0                    
            player_distribution[pid][week][h] += chunk #n(pid)x2x24
            remaining -= chunk
            mi = 0
            h += 1
            if h == 24:
                h = 0
                wd = (wd + 1) % 7
    print(f'dealing:{len(pids)}')
    return player_distribution
 
```

```python
platform_distribution = {} # 3x2x24
all_player = {pid: [[0] * 24, [0] * 24] for pid in pid_list} #the average time in this hour for all players pid x 2 x 24
for name, data in [('steam', steam), ('xbox', xbox), ('nintendo', nintendo)]:
    one = {pid: [[0] * 24, [0] * 24] for pid in pid_list}#prepara for the empty dict 
    one = get_play_distribution_in_one_platform(one, data) #'pid' x 2 x 24
    #the period of each player
    col = 'approximate_session_start' if 'approximate_session_start' in data.columns else 'session_start'
    first = data.groupby('pid')[col].min()
    last = data.groupby('pid')[col].max()
    days = {}
    for pid in first.index:#the key
        y1, m1, d1, h1, mi1 = parse_time(first[pid])
        y2, m2, d2, h2, mi2 = parse_time(last[pid])
        days[pid] = (datetime.date(y2, m2, d2) - datetime.date(y1, m1, d1)).days + 1

    total = [[0] * 24, [0] * 24]
    for pid in one:
        if pid not in days:
            continue
        for i in range(2):
            for j in range(24):
                v = one[pid][i][j] / days[pid] 
                total[i][j] += v
                all_player[pid][i][j] += v
    platform_distribution[name] = total
steam_platform = platform_distribution['steam']
xbox_platform = platform_distribution['xbox']
nintendo_platform = platform_distribution['nintendo']
platform_distribution
```

```python
age_group_list = []
for a in age_list:
    if str(a) == 'nan':
        age_group_list.append(None)
    elif int(a) <= 24:
        age_group_list.append('18-24')
    elif int(a) <= 29:
        age_group_list.append('25-29')
    else:
        age_group_list.append('30+')

feature_lists = {
    'country': country_list,
    'gender': gender_list,
    'age': age_group_list,
    'marital_status': marital_status_list,
    'employment': employment_list,
    'care_children': care_children_list,
}
```

```python
def get_distribution_from_geography(all_player, feature):
    if feature not in feature_lists:
        print(f"Feature '{feature}' is not supported.")
        return None, None
    this_feature = feature_lists[feature]
    count = {}
    #get top3 feature
    for _ in range(len(pid_list)):
        v = this_feature[_]
        if v is None or str(v) == 'nan':
            continue
        count[v] = count.get(v, 0) + 1
    top3 = sorted(count, key=count.get, reverse=True)[:3]
    result = {c: [[0] * 24, [0] * 24] for c in top3} #3x2x24
    n_player = {c: 0 for c in top3} #the number of players for each feature
    for _ in range(len(pid_list)):
        v = this_feature[_]
        if v not in result:
            continue
        pid = pid_list[_]
        n_player[v] += 1
        for i in range(2):
            for j in range(24):
                result[v][i][j] += all_player[pid][i][j]
    return result, n_player
```

```python
geo_distribution = {}   
geo_count = {}          
for feature in feature_lists:
    geo_distribution[feature], geo_count[feature] = get_distribution_from_geography(all_player, feature)
```

```python
platform_24 = {}
for platform_name in platform_distribution:
    platform_24[platform_name] = []
    for j in range(24):
        platform_24[platform_name].append(platform_distribution[platform_name][0][j] + platform_distribution[platform_name][1][j])

geo_24={}
for feature_name in geo_distribution:
    geo_24[feature_name]={}
    for c in geo_distribution[feature_name]:
        geo_24[feature_name][c] = []
        for j in range(24):
            geo_24[feature_name][c].append(geo_distribution[feature_name][c][0][j] + geo_distribution[feature_name][c][1][j])

```

```python
from bokeh.plotting import figure, show
from bokeh.layouts import gridplot
from bokeh.io import output_notebook
output_notebook()

hours = list(range(24))
dashes = ['solid', 'dashed', 'dotted']
greys = ['black', 'dimgray', 'darkgray']
Y_MAX = 10

no_kid = [0] * 24
no_kid_n = 0
for _ in range(len(pid_list)):
    if str(care_children_list[_]) == 'nan':
        no_kid_n += 1
        for j in range(24):
            no_kid[j] += all_player[pid_list[_]][0][j] + all_player[pid_list[_]][1][j]
geo_24['care_children']['No children'] = no_kid
geo_count['care_children']['No children'] = no_kid_n

platform_n = {'steam': steam['pid'].nunique(), 'xbox': xbox['pid'].nunique(), 'nintendo': nintendo['pid'].nunique()}

panels = {}
p = figure(title='Platform', x_axis_label='Hour (local)', y_axis_label='Minutes per player per day',
           y_range=(0, Y_MAX),
           tooltips=[('hour', '$x{0}'), ('minutes', '$y{0.0}')])
k = 0
for name in platform_24:
    y = []
    for j in range(24):
        y.append(platform_24[name][j] / platform_n[name])
    p.line(hours, y, legend_label=f'{name} (n={platform_n[name]})',
           line_dash=dashes[k], color=greys[k], line_width=2)
    k += 1
p.legend.location = 'top_left'
p.legend.click_policy = 'hide'
p.legend.label_text_font_size = '8pt'
panels['platform'] = p

for feature in ['care_children', 'gender', 'employment']:
    p = figure(title=feature, x_axis_label='Hour (local)', y_axis_label='Minutes per player per day',
               y_range=(0, Y_MAX),
               tooltips=[('hour', '$x{0}'), ('minutes', '$y{0.0}')])
    k = 0
    for c in geo_24[feature]:
        n = geo_count[feature][c]
        y = []
        for j in range(24):
            y.append(geo_24[feature][c][j] / n)
        p.line(hours, y, legend_label=f'{c} (n={n})',
               line_dash=dashes[k], color=greys[k], line_width=2)
        k += 1
    p.legend.location = 'top_left'
    p.legend.click_policy = 'hide'
    p.legend.label_text_font_size = '8pt'
    panels[feature] = p

plots = [panels['platform'], panels['gender'], panels['employment'], panels['care_children']]
grid = gridplot(plots, ncols=2, width=600, height=400)
show(grid)
```

```python
import pandas as pd

if 'session_start' in steam.columns:
    steam_logs = steam
else:
    steam_logs = steam.rename(columns={'approximate_session_start': 'session_start',
                                       'approximate_session_end': 'session_end',
                                       'minutes': 'duration'})
cols = ['pid', 'title_id', 'session_start', 'session_end', 'duration']
gaming_logs = pd.concat([
    xbox[cols].assign(platform='xbox'),
    steam_logs[cols].assign(platform='steam'),
    nintendo[cols].assign(platform='nintendo'),
], ignore_index=True)

m = meta.copy()
m['platform'] = m['platform'].str.lower()                      
m['genre_all'] = m['genres'].fillna(m['xbox_primary_genre'])    
m = m.drop_duplicates(['original_name', 'platform'])[['original_name', 'platform', 'genre_all']]
gaming_logs = gaming_logs.merge(m, left_on=['title_id', 'platform'],
                                right_on=['original_name', 'platform'], how='left')
print('percentage of sessions matched with genre:')
print(gaming_logs.groupby('platform')['genre_all'].apply(lambda s: s.notna().mean()).round(3))


g = gaming_logs.dropna(subset=['genre_all']).copy()
g['genre'] = g['genre_all'].str.split(', ')
g['share'] = g['duration'] / g['genre'].str.len()
g = g.explode('genre')


merge_map = {
    'Role-playing (RPG)': 'RPG',          
    'Role Playing': 'RPG',                
    'Action + Adventure': 'Adventure',    
}
g['genre'] = g['genre'].replace(merge_map)

genre_minutes = g.groupby('genre')['share'].sum().sort_values(ascending=False)
print('\ntotal play time by genre (minutes):')
print(genre_minutes.head(10).round(0))

top5 = genre_minutes.head(5).index.tolist()
g['genre_top'] = g['genre'].where(g['genre'].isin(top5), 'Other')
cats = top5 + ['Other']


person = g.pivot_table(index='pid', columns='genre_top', values='share', aggfunc='sum', fill_value=0)
person = person.reindex(columns=cats, fill_value=0)
person = person.div(person.sum(axis=1), axis=0) * 100
person['gender'] = person.index.map(intake.set_index('pid')['gender'])

genders = ['Man', 'Woman', 'Non-binary']
person = person[person['gender'].isin(genders)]
by_gender = person.groupby('gender')[cats].mean().reindex(genders)
n_gender = person.groupby('gender').size().reindex(genders)
print('\neach gender\'s average genre share (%):')
print(by_gender.round(1).assign(n=n_gender))

from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource
from bokeh.palettes import Colorblind
from bokeh.io import output_notebook
output_notebook()

labels = [f'{gd} (n={n_gender[gd]})' for gd in genders]
data = {'gender': labels}
for c in cats:
    data[c] = by_gender[c].tolist()
source = ColumnDataSource(data)

p = figure(y_range=labels[::-1], x_range=(0, 100), width=900, height=300,
           title='Share of play time by genre (average player in each gender)',
           x_axis_label='% of the player\'s play time', toolbar_location=None,
           tooltips='$name: @$name{0.0}%')
p.hbar_stack(cats, y='gender', height=0.6, source=source,
             color=Colorblind[len(cats)],                              
             hatch_pattern=[' ', '/', '.', 'x', '-', '\\'][:len(cats)],     
             legend_label=cats)
p.legend.orientation = 'horizontal'
p.add_layout(p.legend[0], 'below')
p.ygrid.grid_line_color = None
show(p)
```
